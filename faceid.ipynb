```python
#standard dependencies
import cv2
import os
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
#tensorflow dependencies
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense, Layer
```

```python
positive_images_path = os.path.join('data', 'positive')
negative_images_path = os.path.join('data', 'negative')
anchor_images_path = os.path.join('data', 'anchor')
```

```python
#make the directories
os.makedirs(positive_images_path, exist_ok=True)
os.makedirs(negative_images_path, exist_ok=True)
os.makedirs(anchor_images_path, exist_ok=True)
```

```python
# Move LFW images to the dataset folder

for directory in os.listdir('lfw'):
    
    directory_path = os.path.join('lfw', directory)

    for file in os.listdir(directory_path):
        
        EX_PATH = os.path.join(directory_path, file)
        NEW_PATH = os.path.join(negative_images_path, file)
        
        os.replace(EX_PATH, NEW_PATH)
```

```python
#Establish a connection to web cam for image collection
cap = cv2.VideoCapture(1)
while cap.isOpened():
    ret, frame = cap.read()

    #show image in screen
    cv2.imshow('Image Collection', frame)

    #breaking
    if cv2.waitKey(1) & 0xFF == ord ('q'):
     break
#release
cap.release()
cv2.destroyAllWindows()
```
